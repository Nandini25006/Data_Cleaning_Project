```python
import pandas as pd

df = pd.read_csv('titanic.csv')
df.head()
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)

df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

df.drop_duplicates(inplace=True)

print("Data cleaned successfully!")
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

df['Survived'].value_counts().plot(kind='bar')

plt.title('Survival Count')
plt.xlabel('Survived (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()
```

```python
plt.hist(df['Age'])

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```

Conclusion:

1. The dataset was cleaned by handling missing values and removing duplicates.
2. Most passengers did not survive compared to those who survived.
3. The age distribution shows most passengers were adults.
4. The gender distribution shows the proportion of male and female passengers.
5. Data visualization helped identify patterns in the Titanic dataset.
